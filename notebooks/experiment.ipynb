## Understanding ConfigBox

```python
example = {
    "key1":"value1",
    "key2":"value2",
}
```

```python
example["key1"]
```

```python
example.key1
```

```python
from box import ConfigBox
```

```python
example = ConfigBox({
    "key1":"value1",
    "key2":"value2",
})
example.key1
```

## Understanding Ensure

```python
def get_product(x: int, y: int) -> int:
    return x*y
```

```python
get_product(2, 3)
```

```python
get_product(2, "3")
```

```python
from ensure import ensure_annotations

@ensure_annotations
def get_product(x: int, y: int) -> int:
    return x*y
```

```python
get_product(2, 3)
```

```python
get_product(2, "3")
```
